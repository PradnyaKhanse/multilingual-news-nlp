```python
# Step 2: Environment configuration and imports
%load_ext autoreload
%autoreload 2

import sys
import os
import torch
from pathlib import Path
import pandas as pd
from IPython.display import Image, display, Markdown

print(f"PyTorch Version: {torch.__version__}")
print(f"CUDA Available:  {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU Device:      {torch.cuda.get_device_name(0)}")

# Setup project path
PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.config import CONFIG
from src.classify import evaluate_classifier_performance
from src.summarize import evaluate_summarizer_performance
from src.translate import evaluate_translation_performance
from src.evaluate_all import run_full_evaluation
```

## 1. News Topic Classification Evaluation (Fine-Tuned XLM-RoBERTa)
Evaluates the fine-tuned classifier on the held-out test split (~5,560 articles) and renders the Confusion Matrix.

```python
# Evaluate fine-tuned classifier (automatically resolves checkpoint directory on disk)
class_results = evaluate_classifier_performance(save_plot=True)

# Display confusion matrix inline
if class_results.get("plot_path") and Path(class_results["plot_path"]).exists():
    display(Image(filename=class_results["plot_path"]))
```

## 2. Abstractive Summarization Evaluation (XL-Sum Hindi Benchmark)
Evaluates on the real **XL-Sum Hindi test set** (100 articles) and computes **ROUGE-1**, **ROUGE-2**, and **ROUGE-L** scores.

```python
# Evaluate summarizer on 100 XL-Sum Hindi test articles
sum_results = evaluate_summarizer_performance(
    sample_size=100,
    batch_size=8
)

# Display ROUGE chart
rouge_plot = CONFIG.outputs_dir / "plots" / "summarization_rouge_chart.png"
if rouge_plot.exists():
    display(Image(filename=str(rouge_plot)))
```

## 3. Multilingual Translation Evaluation (Hindi → English & Hindi → Marathi)
Evaluates on 100 test translation pairs and computes **BLEU** and **SacreBLEU** scores.

```python
# Evaluate cross-lingual translation on 100 benchmark test pairs
trans_results = evaluate_translation_performance(
    sample_size=100,
    batch_size=8
)

# Display BLEU chart
bleu_plot = CONFIG.outputs_dir / "plots" / "translation_bleu_chart.png"
if bleu_plot.exists():
    display(Image(filename=str(bleu_plot)))
```

## 4. Consolidated System Evaluation (All-In-One Execution)

```python
# Execute master consolidated evaluation across all 3 components
full_report = run_full_evaluation(
    sum_samples=100,
    trans_samples=100
)

# Render 3-panel comprehensive benchmark dashboard chart
dashboard_path = CONFIG.outputs_dir / "plots" / "pipeline_comprehensive_evaluation.png"
if dashboard_path.exists():
    display(Image(filename=str(dashboard_path)))
```
