## 1. Download and Prepare the Real Dataset

By default, the training script uses a tiny 15-example sample dataset bundled in the repo. For actual training, we need to download the full **Inshorts Hindi News dataset** from Kaggle (`shivamtaneja2304/inshorts-dataset-hindi`).

We use the `kagglehub` library, which downloads datasets directly without requiring you to manually upload `kaggle.json` API keys in Colab (it handles public datasets transparently).

```python
import kagglehub
import os
import pandas as pd
from src.preprocess import load_and_clean_classification_data, prepare_classification_splits, save_processed_datasets

# Download the latest version of the dataset
print("Downloading dataset from Kaggle...")
dataset_path = kagglehub.dataset_download("shivamtaneja2304/inshorts-dataset-hindi")
print(f"Dataset downloaded to: {dataset_path}")

# Find the CSV file in the downloaded folder
csv_files = [f for f in os.listdir(dataset_path) if f.endswith('.csv')]
raw_csv_path = os.path.join(dataset_path, csv_files[0])
print(f"Found CSV: {raw_csv_path}")

# Load and rename columns so the pipeline understands it
# The Kaggle dataset uses ['Headline', 'Content', 'Category', 'Date']
# Our pipeline expects 'article' and 'category'
df_raw = pd.read_csv(raw_csv_path)
df_raw = df_raw.rename(columns={"Content": "article", "Category": "category"})

# Save it temporarily in our raw data folder
real_data_csv = str(PROJECT_ROOT / "data" / "raw" / "inshorts_hindi_real.csv")
os.makedirs(os.path.dirname(real_data_csv), exist_ok=True)
df_raw.to_csv(real_data_csv, index=False)
print(f"Saved formatted dataset to: {real_data_csv}\n")

# Clean text, filter invalid categories, and split into train/val/test
df_clean = load_and_clean_classification_data(real_data_csv)
print(f"Cleaned dataset size: {len(df_clean)} articles")
train_df, val_df, test_df = prepare_classification_splits(df_clean)
dataset_paths = save_processed_datasets(train_df, val_df, test_df)
print("\nReady for training!")
```

## 2. Fine-Tune XLM-RoBERTa Model with Hugging Face Trainer

```python
# Execute fine-tuning using the newly processed splits
trained_model_path = train_classifier(
    train_csv=str(dataset_paths['train']),
    val_csv=str(dataset_paths['val']),
    model_name="xlm-roberta-base",
    num_epochs=3,
    batch_size=16,
    learning_rate=2e-5
)
print(f"Model saved at: {trained_model_path}")
```

## 3. Evaluate Model Performance & Generate Confusion Matrix

```python
# Evaluate using the test split and fine-tuned model checkpoint
eval_results = evaluate_classifier_performance(
    test_csv=str(dataset_paths['test']),
    model_path=trained_model_path,
    save_plot=True
)
print(f"Accuracy: {eval_results['accuracy'] * 100:.2f}%")
print(f"Macro F1: {eval_results['f1_macro']:.4f}")
```

## 4. Real-Time Inference on Custom Hindi News

```python
test_article = """
भारतीय अंतरिक्ष अनुसंधान संगठन (इसरो) ने आज नए संचार उपग्रह का सफल परीक्षण किया।
वैज्ञानिकों ने इसे देश की तकनीकी प्रगति के लिए महत्वपूर्ण उपलब्धि बताया।
"""

pred = predict_category(test_article, model_path=trained_model_path)
print("Predicted Category:", pred["category"], f"({pred['category_hi']})")
print("Confidence:", f"{pred['confidence']:.2%}")
print("All Probabilities:", pred["probabilities"])
```
