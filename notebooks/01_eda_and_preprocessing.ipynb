# Notebook 01: Exploratory Data Analysis & Preprocessing for Hindi News

This notebook performs exploratory data analysis (EDA) and Devanagari text normalization for the Multilingual Hindi News NLP System.

### Objectives:
1. Inspect category distributions and article length characteristics.
2. Demonstrate Devanagari Unicode NFC normalization, Nukta unification, and noise cleaning.
3. Stratify and save train/validation/test splits.

```python
import os
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Add project root to sys.path
PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.config import CONFIG
from src.preprocess import clean_hindi_text, load_and_clean_classification_data, prepare_classification_splits, save_processed_datasets
```

## 1. Load and Inspect Raw Classification Data

```python
df_raw = pd.read_csv(CONFIG.raw_data_dir / "sample_hindi_news.csv")
print(f"Total raw records: {len(df_raw)}")
df_raw.head()
```

## 2. Devanagari Normalization & Text Cleaning Demonstration

```python
sample_dirty = "<p>संसद के विशेष सत्र में <b>विपक्ष</b> ने भारी हंगामा किया! https://news.example.com </p>"
sample_cleaned = clean_hindi_text(sample_dirty)
print("Dirty:", sample_dirty)
print("Clean:", sample_cleaned)
```

## 3. Class Distribution & Article Length Analysis

```python
df_clean = load_and_clean_classification_data()
df_clean['char_len'] = df_clean['clean_text'].apply(len)
df_clean['word_len'] = df_clean['clean_text'].apply(lambda x: len(x.split()))

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.countplot(data=df_clean, x='category', palette='viridis')
plt.title('Category Distribution')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.histplot(df_clean['word_len'], bins=10, kde=True, color='teal')
plt.title('Word Count Distribution')
plt.tight_layout()
plt.show()
```

## 4. Stratified Dataset Splitting

```python
train_df, val_df, test_df = prepare_classification_splits(df_clean)
saved_paths = save_processed_datasets(train_df, val_df, test_df)
print("Dataset splits prepared successfully.")
```
